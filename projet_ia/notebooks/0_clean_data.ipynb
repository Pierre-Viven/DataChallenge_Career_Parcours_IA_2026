# Nettoyage des données 

```python
import pandas as pd
import numpy as np
import ast
```

### Mise en forme de x_train et x_test

```python
x_train_brut = pd.read_csv('../data/x_train.csv')
x_test_brut = pd.read_csv('../data/x_test.csv')
print("x_train :")
display(x_train_brut.head())
print("x_test :")
display(x_test_brut.head())


employee_list_train = x_train_brut['employee embedding'].apply(ast.literal_eval).tolist()
company_list_train = x_train_brut['company embedding'].apply(ast.literal_eval).tolist()

employee_list_test = x_test_brut['employee embedding'].apply(ast.literal_eval).tolist()
company_list_test = x_test_brut['company embedding'].apply(ast.literal_eval).tolist()

x_train_employee = np.array(employee_list_train)
x_train_company = np.array(company_list_train)

x_test_employee = np.array(employee_list_test)
x_test_company = np.array(company_list_test)

x_train_embedding = np.hstack((x_train_employee, x_train_company))
x_test_embedding  = np.hstack((x_test_employee, x_test_company))



print("Forme finale de la matrice x_train_embedding  :\n", x_train_embedding.shape,"\n")
display(x_train_embedding [:2])


print("Forme finale de la matrice x_test_embedding  :\n", x_test_embedding.shape,"\n")
display(x_test_embedding [:2])

```

```python

# 1. Obtenir le nombre de colonnes pour chaque type d'embedding
n_train_employee_features = x_train_employee.shape[1]
n_train_company_features = x_train_company.shape[1]

n_test_employee_features = x_test_employee.shape[1]
n_test_company_features = x_test_company.shape[1]

# 2. Générer les noms de colonnes explicites
employee_train_cols = [f'employee_feature_{i}' for i in range(n_train_employee_features)]
company_train_cols = [f'company_feature_{i}' for i in range(n_train_company_features)]

employee_test_cols = [f'employee_feature_{i}' for i in range(n_test_employee_features)]
company_test_cols = [f'company_feature_{i}' for i in range(n_test_company_features)]

# Combinaison des noms de colonnes
all_feature_cols_train = employee_train_cols + company_train_cols
all_feature_cols_test = employee_test_cols + company_test_cols

# 3. Créer le DataFrame Pandas
x_train = pd.DataFrame(x_train_embedding, columns=all_feature_cols_train)
x_test = pd.DataFrame(x_test_embedding, columns=all_feature_cols_test)

# 4. Ajouter la colonne ID au tout début
ids_train = x_train_brut['id'].values
x_train.insert(0, 'id', ids_train)

ids_test = x_test_brut['id'].values
x_test.insert(0, 'id', ids_test)

# Afficher les 5 premières lignes et quelques colonnes
display(x_train.head())
display(x_test.head())
```

```python
fichier_sortie = "../data/x_train_clean.csv"
x_train.to_csv(fichier_sortie, index=False)
print(f"Le fichier propre a été sauvegardé avec succès sous : {fichier_sortie}")

fichier_sortie = "../data/x_test_clean.csv"
x_test.to_csv(fichier_sortie, index=False)
print(f"Le fichier propre a été sauvegardé avec succès sous : {fichier_sortie}")
```

### Mise en forme de y_train

```python

# 1. Charger le CSV brut en ignorant la première colonne d'index anonyme
y_train_brut = pd.read_csv('../data/y_train.csv', index_col=0)

# 2. Définir la hiérarchie ordonnée des postes
mapping_positions = {
    "Assistant": 0,
    "Executive": 1,
    "Manager": 2,
    "Director": 3
}

# 3. Créer la colonne cible encodée
y_train_brut['target'] = y_train_brut['position'].map(mapping_positions)

# 4. Garder uniquement les colonnes utiles (id, position brute, target)
y_train_clean = y_train_brut[['id', 'position', 'target']]

# 5. Sauvegarder le fichier propre sans l'index Pandas
y_train_clean.to_csv('../data/y_train_clean.csv', index=False)

display(y_train_clean.head(10))
```
