```python
import pandas as pd
import numpy as np
import ast
```

## Mise en forme du data frame X_train

```python
X_train_df = pd.read_csv('../data/x_train.csv')
print(X_train_df.head())

emp_list = X_train_df['employee embedding'].apply(ast.literal_eval).tolist()
ent_list = X_train_df['company embedding'].apply(ast.literal_eval).tolist()

X_emp = np.array(emp_list)
X_ent = np.array(ent_list)

X_train = np.hstack((X_emp, X_ent))

print("Forme finale de la matrice X_train :", X_train.shape)
print(X_train[:5])
```

```python
import pandas as pd

# 1. Obtenir le nombre de colonnes pour chaque type d'embedding
n_emp_features = X_emp.shape[1]
n_ent_features = X_ent.shape[1]

# 2. Générer les noms de colonnes explicites
emp_cols = [f'emp_feature_{i}' for i in range(n_emp_features)]
company_cols = [f'company_feature_{i}' for i in range(n_ent_features)]

# Combinaison des noms de colonnes
all_feature_cols = emp_cols + company_cols

# 3. Créer le DataFrame Pandas
df_X_train = pd.DataFrame(X_train, columns=all_feature_cols)

# 4. Ajouter la colonne ID au tout début
ids = X_train_df['id'].values
df_X_train.insert(0, 'id', ids)

# Afficher les 5 premières lignes et quelques colonnes
print("Colonnes générées :", df_X_train.columns[:5], "...", df_X_train.columns[-5:])
print(df_X_train.head())
```

```python
fichier_sortie = "../data/x_train_clean.csv"
df_X_train.to_csv(fichier_sortie, index=False)
print(f"Le fichier propre a été sauvegardé avec succès sous : {fichier_sortie}")
```

## Mise en forme dataframe y_train

```python
import pandas as pd

# 1. Charger le CSV brut en ignorant la première colonne d'index anonyme
y_df = pd.read_csv('../data/y_train.csv', index_col=0)

# 2. Définir la hiérarchie ordonnée des postes
mapping_positions = {
    "Assistant": 0,
    "Executive": 1,
    "Manager": 2,
    "Director": 3
}

# 3. Créer la colonne cible encodée
y_df['target'] = y_df['position'].map(mapping_positions)

# 4. Garder uniquement les colonnes utiles (id, position brute, target)
y_clean = y_df[['id', 'position', 'target']]

# 5. Sauvegarder le fichier propre sans l'index Pandas
y_clean.to_csv('../data/y_train_clean.csv', index=False)

print(y_clean.head(10))
```
